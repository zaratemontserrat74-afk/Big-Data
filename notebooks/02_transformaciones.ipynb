```python
df.set_index(["code", "year"]) 
# se van a usar dos columnas (code y year) como índice compuesto
```

```python
df.loc[df["code"].isnull()] 
# selecciona todas las filas donde la columna code tiene valores nulos (NaN).
```

```python
df.loc[df["code"].notnull(),["entity","year","life_expectancy_0"]].head() 
# df["code"].notnull() → genera una máscara booleana que es True en las filas donde la columna code no está vacía (no es NaN).
# df.loc[condición, columnas] → selecciona las filas que cumplen la condición y, además, solo muestra las columnas especificadas (entity, year, life_expectancy_0).
```

```python
df.dropna(subset=["code"], inplace=True)
# Elimina todas las filas donde la columna code está vacía (NaN).
# subset=["code"] → indica que solo se revisa la columna code para decidir si se borra la fila.
# inplace=True → el cambio se aplica directamente sobre el df, no devuelve una copia.

df.isnull().sum()
#Cuenta cuántos valores nulos (NaN) hay en cada columna.
#Devuelve una serie con el nombre de la columna y el número de valores faltantes.
```

```python
df.shape
# devuelve una tupla (pareja) con el número de filas y columnas del df
```

```python
#Consulta
resultado = (
    df.groupby("entity")["life_expectancy_0"] #Agrupa el df por la columna entity 
    .mean() # Calcula el promedio de life_expectancy_0 para cada grupo.
    .sort_values(ascending=True)  # Ordena esos promedios de menor a mayor.
)
resultado.head(10) # Muestra las primeras 10 entidades con menor esperanza de vida promedio.
```

```python
df_1980 = df[(df["year"] >= 1910) & (df["year"] <= 1910) ] ## Seleccionar un año del periodo historico
resultado = (
    df_1980.groupby("entity")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)
resultado.head(10)
```
