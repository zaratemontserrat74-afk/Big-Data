#Hecho en clase
Indexadores iloc y loc 

Loc (significa location utiliza etuiquetas)--> seleccionar por etiquetas/nombres
- df.loc[filas, columnas]
- Ej: df.iloc[10] --> Regresa la fila que ocupa la posición 10
- Ej: df.iloc[:,3] --> Regresa todas las filas de la columna que ocupa la posión 3
- Se ocupan valores numericos  pero tambien texto
- Busca el valos en índice en las filas

Iloc  (significa integer location)--> seleccionar por posiciones numericas 
- df.iloc[filas, columnas]
- Ej: df.iloc[10] --> Regresa la fila cuya etiqueta de índice es igual a 10
- Ej: df.iloc[:,"Year"] --> Regresa todas las filas de la colmna cuyo nombre es Year
- Se ocupan valores numericos nada más
- Busaca la posicíon en las filas

```python
df.loc[:,["Entity", "Year"]] #:, es para las filas
```

```python
df.iloc[30:35,0:12] 
```
