```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import requests
```

```python
# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/children-born-per-woman.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/children-born-per-woman.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head()
```

```python
df.describe()
```

```python
df.shape
```

```python
# Crea un subconjunto del df original que contiene solo los registros de 

# México.
mexico = df[df["entity"]=="Mexico"]

# Jápon
japon = df[df["entity"]=="Japan"]
```

```python
# Qué tipo de objeto es la variable Mexico.
type(mexico)
```

```python
mexico.head()
```

```python
japon.head()
```

```python
anios_mexico = mexico["year"].to_numpy()
tasa_mexico = mexico["fertility_rate_hist"].to_numpy()
```

```python
anios_japon = japon["year"].to_numpy()
tasa_japon = japon["fertility_rate_hist"].to_numpy()
```

```python
type(anios_mexico)
```

```python
type(anios_japon)
```

```python
promedio_mexico = np.mean(tasa_mexico)
maximo_mexico = np.max(tasa_mexico)

minimo_mexico = np.min(tasa_mexico)
desv_std_mexico = np.std(tasa_mexico)


print("Promedio México:", promedio_mexico)
print("Máximo México:", maximo_mexico)
print("Mínimo México:", minimo_mexico)
print("Desviación estándar México:", desv_std_mexico)
```

```python
promedio_japon = np.mean(tasa_japon)
maximo_japon = np.max(tasa_japon)

minimo_japon = np.min(tasa_japon)
desv_std_japon = np.std(tasa_japon)

print("Promedio Japón:", promedio_japon)
print("Máximo Jápon:", maximo_japon)
print("Mínimo Jápon:", minimo_japon)
print("Desviación estándar Jápon:", desv_std_japon)
```

```python
plt.figure(figsize=(10,5))
```

```python
#Genera una gráfica México
plt.plot(
    anios_mexico,   # eje X: lista/serie con los años
    tasa_mexico,    # eje Y: lista/serie con la tasa de natalidad
    label="Mexico"  # etiqueta para la línea en la leyenda
)


plt.title("Evolución de la tasa de natalidad de México")
plt.xlabel("Año")
plt.ylabel("Nacimiento por cada 1,000 habitantes")
plt.grid(False) # cuadrícula
plt.legend()
```

```python
#Genera una gráfica Japon
plt.plot(
    anios_japon,
    tasa_japon,
    label="Japon"
)

plt.title("Evolución de la tasa de natalidad de Japon")
plt.xlabel("Año")
plt.ylabel("Nacimiento por cada 1,000 habitantes")
plt.grid(True)
plt.legend()
```

```python
#Genera una gráfica México y Jápon 
plt.figure(figsize=(10,5))

plt.plot(
    anios_mexico,   
    tasa_mexico,    
    label="Mexico"  
)
plt.plot(
    anios_japon,
    tasa_japon,
    label="Japon"
)

plt.title("Comparación de la tasa de natalidad")
plt.xlabel("Año")
plt.ylabel("Nacimiento por cada 1,000 habitantes")
plt.grid(True)
plt.legend()
```

```python
##Otra forma de estraer los datos, es preparar desde el data frame de pandas
#ahora usaremos una funcion llamada merge()
```
