4. Seleccionar variables

```python
variables_numericas = [
    "population",
    "births",
    "deaths",
    "child_deaths_under_5",
    "health_percentage",
    "gdp",
    "life_expectancy"
]
df_vn = df[variables_numericas].copy()
df_vn.head(3)
```

5. Exploramos las combinaciones

```python
from itertools import combinations
from math import comb

#Calcular el número de combinaciones
n = len(variables_numericas)
total_combinaciones = comb(n,2)

print(f"Números de combinaciones: {total_combinaciones}")
```

```python
#inicializamos el arreglo para guardar los resultados
resultados = []

for variables_x, variables_y in combinations(variables_numericas, 2):
    # se calcula r mediante la funcion corr de pandas
    r = df_vn[variables_x].corr(df_vn[variables_y], method="pearson")

    # se guarda en un arreglo todos los valores obtenidos
    resultados.append([variables_x, variables_y, r, abs(r)])

# el arreglo de resultads se tranforma en df pandas para manipularlo
#Convertir los resultados en un DataFrame
df_resultados = pd.DataFrame(
    resultados,
    columns=["Variable X",
             "Variable Y",
             "Pearson",
             "Intensidad"] #es el valor absoluto de r
)

#Ordenar por intensidad de mayor a menor
df_resultados = df_resultados.sort_values(
    by="Intensidad", #esta es la variable por la cual se ordena
    ascending=False #aplicamos un ordenamiento ascendente
).reset_index(drop=True) # drop=True evita que el indice original se agregue como

display(df_resultados)
```
